### Yacht Hydrodynamics Exploratory Data Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset with descriptive column names
file_path = '/content/yacht_hydrodynamics.data'
cols = [
    "long_pos_buoyancy", "prismatic_coeff", "length_disp_ratio",
    "beam_draught_ratio", "length_beam_ratio", "froude_number",
    "residuary_resistance"
]
df_yacht = pd.read_csv(file_path, sep=r'\s+', header=None, names=cols)

print("=== Dataset Structural Overview ===")
print(f"Dataset Shape: {df_yacht.shape}")
print("\nHead of Dataset:")
display(df_yacht.head())
print("\nTail of Dataset:")
display(df_yacht.tail())
print("\nDataset Info:")
df_yacht.info()
print("\nDataset Descriptive Statistics:")
display(df_yacht.describe())
```

```python
# Q1: Find mean, median, minimum, and maximum for each measurement
print("=== Q1: Mean, Median, Min, and Max ===")
q1_summary = df_yacht.agg(['mean', 'median', 'min', 'max']).T
display(q1_summary)
```

```python
# Q2: Which measurement has the largest standard deviation? (Using NumPy)
print("=== Q2: Largest Standard Deviation ===")
np_array = df_yacht.to_numpy()
std_values = np.std(np_array, axis=0, ddof=1)
std_series = pd.Series(std_values, index=cols)
max_std_col = std_series.idxmax()
max_std_val = std_series.max()
print(std_series.round(4))
print(f"\nThe measurement with the largest standard deviation is '{max_std_col}' with a standard deviation of {max_std_val:.4f}.")
```

```python
# Q3: Find the 10 records with the highest resistance value
print("=== Q3: Top 10 Records with Highest Residuary Resistance ===")
top_10_resistance = df_yacht.sort_values(by='residuary_resistance', ascending=False).head(10)
display(top_10_resistance)
```

```python
# Q4: Compare average resistance across low, medium, and high Froude-number groups
print("=== Q4: Froude-number Group Comparison ===")
# Grouping definition:
# Low: Froude number <= 0.25
# Medium: 0.25 < Froude number <= 0.35
# High: Froude number > 0.35
bins = [0, 0.25, 0.35, 1.0]
labels = ['Low (F<=0.25)', 'Medium (0.25<F<=0.35)', 'High (F>0.35)']
df_yacht['froude_group'] = pd.cut(df_yacht['froude_number'], bins=bins, labels=labels)

group_comparison = df_yacht.groupby('froude_group', observed=True)['residuary_resistance'].agg(['count', 'mean', 'median', 'min', 'max'])
display(group_comparison)
```

```python
# Q5: Check for missing values and duplicates
print("=== Q5: Missing Values and Duplicates ===")
missing_vals = df_yacht[cols].isnull().sum()
dup_count = df_yacht[cols].duplicated().sum()
print("Missing values per column:")
print(missing_vals)
print(f"Total Duplicate Rows: {dup_count}")
```

```python
# Q6 & Visualizations
print("=== Q6: Visualizations ===")

# Ensure 'froude_group' is created dynamically within this cell if not already present
if 'froude_group' not in df_yacht.columns:
    bins = [0, 0.25, 0.35, 1.0]
    labels = ['Low (F<=0.25)', 'Medium (0.25<F<=0.35)', 'High (F>0.35)']
    df_yacht['froude_group'] = pd.cut(df_yacht['froude_number'], bins=bins, labels=labels)

# Chart 1: Scatter plot of Froude Number vs Residuary Resistance
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_yacht, x='froude_number', y='residuary_resistance', alpha=0.7, color='teal')
plt.title('Residuary Resistance vs. Froude Number')
plt.xlabel('Froude Number')
plt.ylabel('Residuary Resistance')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# Chart 2: Bar plot comparing Mean Residuary Resistance across Froude-number groups
plt.figure(figsize=(8, 5))
sns.barplot(data=df_yacht, x='froude_group', y='residuary_resistance', errorbar=None, palette='Blues_d', hue='froude_group', legend=False)
plt.title('Mean Residuary Resistance by Froude-Number Group')
plt.xlabel('Froude Group')
plt.ylabel('Mean Residuary Resistance')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Three Key Observations

1. **Strong Non-linear Trend**: The scatter plot shows that residuary resistance exhibits an exponential-like upward trend relative to the Froude number, particularly surging when Froude numbers exceed 0.30.
2. **Dramatic Performance Differences**: Group analysis highlights that the average residuary resistance rises dramatically from ~0.78 for the Low Froude-number group to ~30.56 for the High Froude-number group.
3. **Data Completeness and Variance**: The dataset has clean integrity with 0 missing values or duplicate rows. Statistically, 'residuary_resistance' exhibits the highest variation by far (standard deviation of ~15.16), while the 'prismatic_coeff' is extremely tightly packed with an std of just ~0.023.
